# Exploring the Airbnb listings data

Before running the Evidently drift report, let's look at the two snapshots we're comparing — Airbnb listings from March 2025 and November 2025 — and get a feel for the columns we're tracking.

```python
import pandas as pd

REFERENCE_PATH = "data/listings_2025_march.csv"
CURRENT_PATH = "data/listings_2025_november.csv"

FEATURES = [
    "price",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365",
    "room_type",
    "neighbourhood_group_cleansed",
]


def load_and_clean(path: str) -> pd.DataFrame:
    df = pd.read_csv(path, low_memory=False)
    df = df[FEATURES].copy()
    df["price"] = (
        df["price"].astype(str).str.replace(r"[\$,]", "", regex=True).astype(float)
    )
    return df.dropna()


df_ref = load_and_clean(REFERENCE_PATH)
df_cur = load_and_clean(CURRENT_PATH)

print(f"Reference (2025 March)   : {df_ref.shape}")
print(f"Current   (2025 November): {df_cur.shape}")
```
